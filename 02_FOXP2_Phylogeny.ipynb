```python
#@markdown We decided to experiment with Phylo a bit before getting pair distances.

from Bio import Phylo

# Reading our Newick file and getting species names + protein ID's
tree = Phylo.read("FOXP2ProteinTree.nwk", "newick")


protein = [leaf.name for leaf in tree.get_terminals()]
species = [item.split("_", 1)[1] for item in protein]

print(species)
print(protein)
print("Num. Of Terminal Clades:", len(tree.get_terminals()))
```

```python
#@markdown We wanted to compare every possible pair of FOXP2 sequences to measure how evolutionarily distant they are, so we generated all unique sequence pairs and calculated the tree-based distance between each pair.

import itertools

def terminal_pair_dists(tree_obj):

    def generate_pairs(iterable):
      return itertools.combinations(iterable, 2)

    return [
        (pair[0].name, pair[1].name, tree_obj.distance(*pair))
        for pair in generate_pairs(tree_obj.find_clades(terminal = True))
        ]

distances = terminal_pair_dists(tree)

for p1, p2, dist in distances:
    s1 = species[protein.index(p1)]
    s2 = species[protein.index(p2)]
    print("Number of comparisons:", len(distances))
```

```python
#@markdown Next, we organized the sequence comparisons into a table with the protein IDs, species, and distances, and we extracted that information from each comparison then stored it in a DataFrame.

import pandas as pd

records = []

for p1, p2, dist in distances:
  full1 = p1.split("|")[-1]
  full2 = p2.split("|")[-1]

  protein1, species1 = full1.rsplit("_", 1)
  protein2, species2 = full2.rsplit("_", 2)

  records.append({
      "Protein1": protein1,
      "Protein2": protein2,
      "Species1": species1,
      "Species2": species2,
      "Distance": dist
  })

distance_df = pd.DataFrame(records)
print(distance_df.shape)
distance_df.head()
```

```python
distance_df.to_csv("Distances.csv", index=False)
```

```python
#@markdown  We wanted to see which species had more similar or different FOXP2 sequences, so we grouped the comparisons by species pair, calculated the median distance for each pair, and displayed the 10 smallest and largest median distances.

species_pair_df = (
    distance_df
    .groupby(["Species1", "Species2"])["Distance"]
    .median()
    .reset_index()
    .rename(columns={"Distance": "MedianDistance"})
)

print(species_pair_df.shape)

print("Smallest Distances:")
display(species_pair_df.sort_values("MedianDistance").head(10))
print("Largest Distances:")
display(species_pair_df.sort_values("MedianDistance", ascending=False).head(10))
```

```python
species_pair_df.to_csv("SpeciesPair.csv", index=False)
```

**Rooting Tree**

```python
#@markdown We wanted to focus on a group of species with many biological similarities, so we chose primates as our ingroup because they share many traits, making their FOXP2 relationships interesting to investigate. We chose mouse as the outgroup because it is outside the primate group and could provide a reference point for the evolutionary relationships within the primates.

ingroup = {"HUMAN", "PANTR", "PANPR", "GORGO", "PONAB", "PONPY"}
outgroup = {"MOUSE", "MUSCR"}

ingroup_tips = [
    tip for tip in tree.get_terminals()
    if tip.name and any(tip.name.endswith("_" + species) for species in ingroup)
]

outgroup_tips = [
    tip for tip in tree.get_terminals()
    if tip.name and any(tip.name.endswith("_" + species) for species in outgroup)
]

print("Ingroup sequences:", len(ingroup_tips))
print("Outgroup sequences:", len(outgroup_tips))
```

```python
mouse_ancestor = tree.common_ancestor(outgroup_tips)
tree.root_with_outgroup(mouse_ancestor)
```

```python
print("Rooted tree:")
print(tree.root)
print("\nRoot children:")

for child in tree.root.clades:
    print("Size:", len(child.get_terminals()))
    print("Example:", child.get_terminals()[0].name)
```
